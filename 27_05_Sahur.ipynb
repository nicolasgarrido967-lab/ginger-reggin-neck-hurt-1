```python
#jogo menu inteligente

# criar um sistema que funciona até o usuario escolher sair
# 1 par/impar
# 2 positivo negativo
# 3 maior
# 4 sair

opcao = 0

while opcao != 4:
  print("\n--- MENU ---")
  print("1 - Par/Ímpar")
  print("2 - Positivo/Negativo")
  print("3 - Maior")
  print("4 - Sair")
  print("------------")

  try:
    opcao = int(input("Digite a opção desejada: "))

    if opcao == 1:
      n1 = int(input("Digite um número para verificar se é par ou ímpar: "))
      if n1 % 2 == 0:
        print(f"O número {n1} é PAR.")
      else:
        print(f"O número {n1} é ÍMPAR.")
    elif opcao == 2:
      n1 = int(input("Digite um número para verificar se é positivo ou negativo: "))
      if n1 > 0:
        print(f"O número {n1} é POSITIVO.")
      elif n1 < 0:
        print(f"O número {n1} é NEGATIVO.")
      else:
        print(f"O número {n1} é ZERO.")
    elif opcao == 3:
      n1 = int(input("Digite o primeiro número: "))
      n2 = int(input("Digite o segundo número: "))
      if n1 > n2:
        print(f"O número {n1} é MAIOR que {n2}.")
      elif n2 > n1:
        print(f"O número {n2} é MAIOR que {n1}.")
      else:
        print(f"Os números {n1} e {n2} são IGUAIS.")
    elif opcao == 4:
      print("Saindo do programa. Até mais!")
    else:
      print("Opção inválida. Por favor, escolha uma opção entre 1 e 4.")
  except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro para a opção.")
```

#urna eletronica
#crir um programa de votaçao com 3 candidatos
#o sistema deve repetir até encerrar a votaçao
#contar votos, brancos e nulos
#ao encerrar, mostrar:
#total de votos, brancos e nulos e o vencedor
#mostrar se sitver empate e se nao houver votos validos mostrar:nenhum candidato recebeu votos

```python
# urna eletronica
# Inicializa contadores de votos
votos_candidato1 = 0
votos_candidato2 = 0
votos_candidato3 = 0
votos_brancos = 0
votos_nulos = 0
total_votos_registrados = 0

print("--- Urna Eletrônica ---")
print("Candidatos:")
print("  1 - Candidato A")
print("  2 - Candidato B")
print("  3 - Candidato C")
print("  0 - Voto em Branco")
print("  Qualquer outro número inteiro para Voto Nulo")
print("  Digite 99 para encerrar a votação a qualquer momento.")

while True:
    try:
        voto = input("Digite seu voto: ")

        if voto == '99':
            print("Votação encerrada.")
            break

        voto_int = int(voto)

        if voto_int == 1:
            votos_candidato1 += 1
        elif voto_int == 2:
            votos_candidato2 += 1
        elif voto_int == 3:
            votos_candidato3 += 1
        elif voto_int == 0:
            votos_brancos += 1
        else:
            votos_nulos += 1

        total_votos_registrados += 1
        print("Voto registrado com sucesso!")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")
        # Considera entradas não numéricas como nulas para o total de votos registrados
        votos_nulos += 1
        total_votos_registrados += 1

print("\n--- Resultado da Votação ---")
print(f"Total de votos registrados: {total_votos_registrados}")
print(f"Votos para Candidato A: {votos_candidato1}")
print(f"Votos para Candidato B: {votos_candidato2}")
print(f"Votos para Candidato C: {votos_candidato3}")
print(f"Votos em Branco: {votos_brancos}")
print(f"Votos Nulos: {votos_nulos}")


total_valid_votes = votos_candidato1 + votos_candidato2 + votos_candidato3

if total_valid_votes == 0:
    print("Nenhum candidato recebeu votos válidos.")
else:
    max_votos = votos_candidato1
    if votos_candidato2 > max_votos:
        max_votos = votos_candidato2
    if votos_candidato3 > max_votos:
        max_votos = votos_candidato3

    vencedores = []
    if votos_candidato1 == max_votos:
        vencedores.append("Candidato A")
    if votos_candidato2 == max_votos:
        vencedores.append("Candidato B")
    if votos_candidato3 == max_votos:
        vencedores.append("Candidato C")

    if len(vencedores) > 1:
        print(f"Houve um empate entre os seguintes candidatos: {', '.join(vencedores)} com {max_votos} votos cada.")
    else:
        print(f"O vencedor é: {vencedores[0]} com {max_votos} votos.")
```
