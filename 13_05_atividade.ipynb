```python
#1-contagem simples, o usuario digite um numero e imprimi do zero até o numero
numero = int(input("Digite um número: "))
for i in range(numero + 1):
    print(i)
```

```python
#2-contagem regrassiva,o usuario digita um numero e o programa imprima até o zero
numero = int(input("Digite um número: "))
for i in range(numero, -1, -1):
  print(i)
```

```python
#5-soma, o usuario digita um numero e o programa soma os numeros do zero até o numero
n = int(input("Digite um número: "))
soma = sum(range(n + 1))
print(f"A soma de 0 até {n} é: {soma}")
```

```python
#7-contar letras, o usuario digita uma palavra e o programa conte as letras da palavra
palavra = input("Digite uma palavra: ")
contador = 0
for letra in palavra:
  contador += 1
print("A palavra tem", contador, "letras.")
```

```python
#8-contar as vogais, o usuario digita uma palavra e o programa conta as vogais
palavra = input("Digite uma palavra: ").lower()

vogais = "aeiou"
contador = 0

for letra in palavra:
    if letra in vogais:
        contador += 1

print(f"A palavra tem {contador} vogais.")
```

```python
#9-maior número digital, o usuario digita 4 numeros e o programa imprima o maior número digitado
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
n3 = int(input("Digite o terceiro número: "))
n4 = int(input("Digite o quarto número: "))
if n1 > n2 and n1 > n3 and n1 > n4:
  print(n1)
if n2 > n1 and n2 > n3 and n2 > n4:
  print(n2)
if n3 > n1 and n3 > n2 and n3 > n4:
  print(n3)
if n4 > n1 and n4 > n2 and n4 > n3:
  print(n4)
```

```python
n = int(input("Digite um número para a pirâmide: "))

for i in range(1, n + 1):
    # print dos espaços pra simetria
    print("  " * (n - i), end="")

    # Print ascending numbers
    for j in range(1, i + 1):
        print(j, end=" ")

    # print dos numeros descendentes
    for j in range(i - 1, 0, -1):
        print(j, end=" ")

    print()
```
