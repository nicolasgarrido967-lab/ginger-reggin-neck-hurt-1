### Exercício 1 — Acessando um elemento

Crie um programa que exiba somente o número 50 da matriz abaixo.

```python
matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# O número 50 está na segunda linha (índice 1) e segunda coluna (índice 1)
print(matriz[1][1])
```

### Exercício 3 — Exibindo uma coluna

Utilize um for para exibir somente os números da primeira coluna da matriz.

```python
matriz = [
    [2, 4, 6],
    [8, 10, 12],
    [14, 16, 18]
]

# Percorre cada linha e exibe o elemento do índice 0 (primeira coluna)
for linha in matriz:
    print(linha[0])
```

### Exercício 4 — Alterando um elemento

Altere o número 20 para 99 e exiba a matriz completa após a alteração.

```python
matriz = [
    [10, 20, 30],
    [40, 50, 60]
]

# O número 20 está na primeira linha (índice 0) e segunda coluna (índice 1)
matriz[0][1] = 99

print(matriz)
```

### Exercício 6 — Guardando uma coluna em uma lista

Crie uma lista vazia chamada coluna. Percorra a matriz, guarde nela os números da segunda coluna e, ao final, exiba a lista.

```python
matriz = [
    [3, 6, 9],
    [12, 15, 18],
    [21, 24, 27]
]

coluna = []

# Percorre a matriz e adiciona à lista o elemento do índice 1 (segunda coluna)
for linha in matriz:
    coluna.append(linha[1])

print(coluna)
```

### Exercício 7 — Exibindo todos os elementos

Utilize dois comandos for para exibir todos os números da matriz, um por linha.

### Exercício 8 — Somando uma coluna

Crie um programa que some todos os números da segunda coluna e exiba o resultado.

```python
matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

soma = 0
for linha in matriz:
    soma += linha[1]

print("Soma da segunda coluna:", soma)
```

### Exercício 9 — Encontrando o maior número

Utilize dois comandos for para percorrer a matriz e encontrar o maior número. Ao final, exiba o resultado.

```python
matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

# Inicializa com o primeiro elemento da matriz
maior = matriz[0][0]

for linha in matriz:
    for elemento in linha:
        if elemento > maior:
            maior = elemento

print("Maior número:", maior)
```

### Exercício 10 — Matriz de notas

Uma matriz armazena as notas de três alunos, sendo cada linha correspondente a um aluno.

```python
notas = [
    [7, 8, 9],
    [6, 5, 8],
    [10, 9, 7]
]

# a) Exiba todas as notas do primeiro aluno.
print("a) Notas do primeiro aluno:", notas[0])

# b) Exiba somente a segunda nota do segundo aluno.
print("b) Segunda nota do segundo aluno:", notas[1][1])

# c) Utilize um for para exibir a terceira nota de cada aluno.
print("c) Terceira nota de cada aluno:")
for linha in notas:
    print(linha[2])

# d) Altere a primeira nota do terceiro aluno para 8 e exiba a matriz atualizada.
notas[2][0] = 8
print("d) Matriz atualizada:", notas)
```
