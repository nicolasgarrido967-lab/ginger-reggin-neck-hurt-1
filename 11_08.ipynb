```python
#exercicio 2
nota1 = 7.5
nota2 = 8.0
nota3 = 6.5


media = (nota1 + nota2 + nota3) / 3


print(f"As notas são: {nota1}, {nota2}, {nota3}")
print(f"A média das notas é: {media:.2f}")
```

```python
#exercicio 3
texto1 = "20"
texto2 = "3.5"
numero = 8

resultado_a = int(texto1) + 10
print(f"a) {texto1} transformado em inteiro + 10: {resultado_a}")

resultado_b = float(texto2) * 2
print(f"b) {texto2} transformado em real * 2: {resultado_b}")

resultado_c = str(numero) + " anos"
print(f"c) {numero} transformado em texto + ' anos': {resultado_c}")
```

```python
#exercicio 4
dobro_15 = 15 * 2
print(f"O dobro de 15 é: {dobro_15}")

metade_90 = 90 / 2
print(f"A metade de 90 é: {metade_90}")

resto_divisao = 17 % 5
print(f"O resto da divisão de 17 por 5 é: {resto_divisao}")

potencia = 2 ** 4
print(f"2 elevado à potência 4 é: {potencia}")

media_numeros = (10 + 8 + 6) / 3
print(f"A média de 10, 8 e 6 é: {media_numeros}")
```

```python
#exercicio 5
idade = 16
nota = 8.5
preco = 45.0

print(f"idade ({idade}) >= 18: {idade >= 18}")
print(f"nota ({nota}) > 7: {nota > 7}")
print(f"preco ({preco}) <= 50: {preco <= 50}")
print(f"idade ({idade}) == 16: {idade == 16}")
print(f"nota ({nota}) != 10: {nota != 10}")
```

```python
#exercicio 6
preco_inicial = 80.00

percentual_desconto = 0.10

valor_desconto = preco_inicial * percentual_desconto
print(f"O valor do desconto é: R$ {valor_desconto:.2f}")

valor_final = preco_inicial - valor_desconto
print(f"O valor final a pagar é: R$ {valor_final:.2f}")

e_menor_que_75 = valor_final < 75.00
print(f"O valor final (R$ {valor_final:.2f}) é menor que R$ 75.00? {e_menor_que_75}")
```
