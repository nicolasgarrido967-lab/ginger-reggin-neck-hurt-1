```python
soma_total = 0
numero_digitado = None

while numero_digitado != 0:
    try:
        numero_digitado = int(input("Digite um número (digite 0 para encerrar): "))
        soma_total += numero_digitado
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

print(f"A soma de todos os números digitados é: {soma_total}")
```

```python
maior_numero = None
numero_digitado_max = None

while numero_digitado_max != 0:
    try:
        numero_digitado_max = int(input("Digite um número (digite 0 para encerrar): "))
        if numero_digitado_max != 0:
            if maior_numero is None or numero_digitado_max > maior_numero:
                maior_numero = numero_digitado_max
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if maior_numero is not None:
    print(f"O maior número digitado foi: {maior_numero}")
else:
    print("Nenhum número válido foi digitado além de 0.")
```

```python
soma_notas = 0
quantidade_notas = 0

while quantidade_notas < 5:
    try:
        nota = float(input(f"Digite a nota {quantidade_notas + 1}: "))
        soma_notas += nota
        quantidade_notas += 1
    except ValueError:
        print("Entrada inválida. Por favor, digite um número para a nota.")

if quantidade_notas > 0:
    media = soma_notas / quantidade_notas
    print(f"A média das notas é: {media:.2f}")
else:
    print("Nenhuma nota foi digitada.")
```

```python
saldo = 1000.00

while True:
    print("\n--- Menu Bancário ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    try:
        opcao = int(input("Escolha uma opção: "))

        if opcao == 1:
            valor_deposito = float(input("Digite o valor a depositar: "))
            if valor_deposito > 0:
                saldo += valor_deposito
                print(f"Depósito de R$ {valor_deposito:.2f} realizado com sucesso.")
            else:
                print("O valor do depósito deve ser positivo.")
        elif opcao == 2:
            valor_saque = float(input("Digite o valor a sacar: "))
            if valor_saque > 0:
                if saldo >= valor_saque:
                    saldo -= valor_saque
                    print(f"Saque de R$ {valor_saque:.2f} realizado com sucesso.")
                else:
                    print("Saldo insuficiente!")
            else:
                print("O valor do saque deve ser positivo.")
        elif opcao == 3:
            print(f"Seu saldo atual é: R$ {saldo:.2f}")
        elif opcao == 4:
            print("Saindo do programa. Obrigado!")
            break
        else:
            print("Opção inválida. Por favor, escolha uma opção de 1 a 4.")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro para a opção e um número para os valores.")
```
