```python
#escrever um programa que recebe 4 notas do aluno e imprime "reprovado", "aprovado" ou "recuperação"
#regras
#usar while
#nota maior ou igual a 7 - aprovado
#nota menor ou igual a 5 - recuperação
#nota menor que 5 - reprovado

sum_notas = 0
contador_notas = 0

while contador_notas < 4:
  nota = float(input(f"Digite a nota {contador_notas + 1}: "))
  sum_notas += nota
  contador_notas += 1

media = sum_notas / 4
print(f"A média do aluno é: {media:.2f}")

if media >= 7:
  print("aprovado")
elif media < 5:
  print("reprovado")
else:
  print("recuperação")
```

```python
saldo = 1000.00
opcao = 0

while opcao != 4:
    print("1 - Ver saldo")
    print("2 - Depositar dinheiro")
    print("3 - Sacar dinheiro")
    print("4 - Sair")

    opcao = int(input("Digite a opção desejada: "))

    if opcao == 1:
      print(f"Seu saldo é igual a {saldo}")

    if opcao == 2:
      valor_deposito = float(input("Digite o valor do seu deposito: "))
      saldo = saldo + valor_deposito
      print(f"")
      print(f"")
      print(f"Você depositou R${valor_deposito}")
      print(f"Seu saldo agora é de: R${saldo}")
      print(f"")
      print(f"")

    if opcao == 3:
      valor_sacar = float(input("Digite o valor que quer sacar: "))
      saldo = saldo - valor_sacar
      print(f"")
      print(f"")
      print(f"Você sacou R${valor_sacar}")
      print(f"Seu saldo agora é de R${saldo}")
      print(f"")
      print(f"")

    if opcao == 4:
      print(f"")
      print(f"OBRIGADO POR USAR ESTE CAIXA ELETRÔNICO!")
```
