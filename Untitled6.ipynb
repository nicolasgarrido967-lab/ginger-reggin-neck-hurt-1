```python
# Cofre digital
# Um cofre possui um código composto por 5 números inteiros.
# Desenvolva um programa que:
# Leia os 5 números utilizando um laço de repetição.
# Ao final informar:
#1 Quantos números são pares.
#2 Quantos números são impares.
#3 O maior número.
#4 A soma de todos os números.
#5 A média dos números.
#6 menor número.
#7 Quantos números são maiores que a média.

numbers = []
for i in range(5):
    num = int(input(f"Digite o {i+1}º número: "))
    numbers.append(num)

pares_count = 0
impares_count = 0
soma = 0
maior_numero = numbers[0]
menor_numero = numbers[0]

for num in numbers:
    soma += num
    if num % 2 == 0:
        pares_count += 1
    else:
        impares_count += 1

    if num > maior_numero:
        maior_numero = num

    if num < menor_numero:
        menor_numero = num

media = soma / len(numbers)

maiores_que_media_count = 0
for num in numbers:
    if num > media:
        maiores_que_media_count += 1

print(f"\nResultados:")
print(f"1. Quantos números são pares: {pares_count}")
print(f"2. Quantos números são ímpares: {impares_count}")
print(f"3. O maior número: {maior_numero}")
print(f"4. A soma de todos os números: {soma}")
print(f"5. A média dos números: {media:.2f}")
print(f"6. O menor número: {menor_numero}")
print(f"7. Quantos números são maiores que a média: {maiores_que_media_count}")
```
